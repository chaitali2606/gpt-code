```python
import torch
from transformers import GPT2Tokenizer, GPT2LMHeadModel
```

```python
tokenizer = GPT2Tokenizer.from_pretrained('gpt2')
model = GPT2LMHeadModel.from_pretrained('gpt2')
model.eval()
```

```python
# Iterate over each row in the dataframe
for i, row in data.iterrows():
    prompt = row["description"]  # replace "column_name_for_prompt" with the actual name of the column in your dataframe
    input_ids = torch.tensor(tokenizer.encode(prompt)).unsqueeze(0)  # encode the prompt
    output = model.generate(input_ids=input_ids, max_length=100, do_sample=True)  # generate text
    generated_text = tokenizer.decode(output[0], skip_special_tokens=True)  # decode the generated text
    print("Generated text for input: ", prompt)
    print(generated_text)
```
